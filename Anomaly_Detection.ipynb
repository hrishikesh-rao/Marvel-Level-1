```python
#Using IQR[Inter-Quartile Ranges] Method Approach
features = [
    'login_duration_min',
    'data_accessed_MB',
    'files_downloaded'
]

for col in features:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col + '_anomaly_IQR'] = (
        (df[col] < lower) | (df[col] > upper)
    )
df.drop(columns=["login_duration_minIQR","data_accessed_MBIQR",	"files_downloadedIQR","login_duration_min_IQR","data_accessed_MB_IQR",	"files_downloaded_IQR","login_duration_min_anoamly_IQR","data_accessed_MB_anoamly_IQR","files_downloaded_anoamly_IQR"], inplace=True)

```

```python
df
```

```python
df['iqr_any_anomaly'] = (
    df['login_duration_min_anomaly_IQR'] |
    df['data_accessed_MB_anomaly_IQR'] |
    df['files_downloaded_anomaly_IQR']
)
df
```

```python
features = ['login_duration_min', 'data_accessed_MB', 'files_downloaded']
X = df[features]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.ensemble import IsolationForest

iso = IsolationForest(
    n_estimators=100,
    contamination=0.05,  # assume ~5% anomalies
    random_state=42
)

df['iso_flag'] = iso.fit_predict(X_scaled)
```

```python
df['iso_anomaly'] = df['iso_flag'] == -1
df
```

```python
df['iso_anomaly'].sum()
```

```python
comparison = df[['iqr_any_anomaly', 'iso_anomaly']]
comparison.value_counts()
```

```python
suspects = df[df['iqr_any_anomaly'] & df['iso_anomaly']]
suspects
```

```python
df['iso_score'] = iso.decision_function(X_scaled)
df.sort_values('iso_score').head(5)
```
