```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("customer_data.csv")
```

```python
df.head()
```

```python
df = df.drop_duplicates()
df
```

```python
df["Name"].str.strip("1234567890._/")
```

```python
df.head(10)
```

```python
df["num"] = df["Name"].str.isnumeric()
df.head(10)
```

```python
df.loc[df["Name"].astype(str).str.isnumeric(), "Name"] = np.nan
df.head(10)
```

```python
df["Gender"] = df["Gender"].str.replace('Femlae','Female')
df["Gender"] = df["Gender"].str.replace('mle','Male')
df.head(10)
```

```python
df = df.drop("num",axis=1)
df
```

```python
df["Country"] = df["Country"].str.replace('Indai','India')
df["Country"] = df["Country"].str.replace('Canda','Canada')
df
```

```python
df["PreferredDevice"]=df["PreferredDevice"].str.replace('dasktop','Desktop')
df["PreferredDevice"]=df["PreferredDevice"].str.replace('moblie','Mobile')
df
```

```python
df["TotalPurchase"] = df["TotalPurchase"].abs()
```

```python
for i in df.index:
    if df.loc[i, "Age"] < 0 or df.loc[i, "Age"] > 100:
        df.loc[i, "Age"] = np.nan
df
```

```python
df = df.dropna(subset=['CustomerID', 'Email', 'TotalPurchase'])
df
```

```python
df = df.reset_index(drop=True)
df
```

```python
df['Age']=df['Age'].fillna(df['Age'].median())
df['Gender']=df['Gender'].fillna('Unknown')
df['PreferredDevice']=df['PreferredDevice'].fillna('Unknown')
df['Country']=df['Country'].fillna('Unknown')
df
```

```python
df['SignupDate'] = pd.to_datetime(df['SignupDate'], errors='coerce')
df['LastLogin'] = pd.to_datetime(df['LastLogin'], errors='coerce')
df
```

```python
df = df[df['Email'].str.contains('invalidemail.com', na=False) == False]
df
```

```python
df['Email'].str.contains(r'^[^@]+@[^@]+\.[^@]+$')
```

```python

valid_email = df['Email'].str.contains(r'^[^@]+@[^@]+\.[^@]+$', na=False)
df.loc[~valid_email, 'Email'] = np.nan

df
```

```python
df.to_csv("cleaned_customer_data.csv", index=False)

from google.colab import files
files.download("cleaned_customer_data.csv")
```
