```python
from faker import Faker
import pandas as pd
import numpy as np
```

```python
fake=Faker()
```

```python
Transaction_ID = []
Account_ID = []
Transaction_Date = []
Transaction_Amount = []
Merchant_Category = []
Transaction_Type = []
Location =  []
is_fraud =[]
n=50
```

```python
for _ in range(n):
    Transaction_ID.append(fake.uuid4())
    Account_ID.append(f"ACC{fake.random_number(digits=8, fix_len=True)}")
    Transaction_Date.append(fake.date_between(start_date='-3y', end_date='today'))
    Transaction_Amount.append(round(np.random.uniform(10, 1000000), 2))
    Merchant_Category.append(fake.random_element(elements=("E-commerce", "Retail", "Medicine", "Tax", "Education")))
    Transaction_Type.append(fake.random_element(elements=("Debit", "Credit")))
    Location.append(fake.city())
    is_fraud.append(fake.boolean(chance_of_getting_true=10))
```

```python
df = pd.DataFrame(zip(Transaction_ID,Account_ID,Transaction_Date,Transaction_Amount,Merchant_Category,Transaction_Type,Location,is_fraud) , columns = ['Transaction_ID','Account_ID','Transaction_Date','Transaction_Amount','Merchant_Category','Transaction_Type','Location','is_fraud'])
```
